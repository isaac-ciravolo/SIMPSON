```python
import pandas as pd

MIN_PA = 1

batting = pd.read_csv("../data/singles_cs.csv")
```

```python
qualified = batting.loc[batting["PA"] >= MIN_PA].copy()

if "1B" not in qualified.columns:
    qualified["1B"] = qualified["H"] - qualified[["2B", "3B", "HR"]].sum(axis=1)

qualified["SIMPSON"] = (
    qualified["1B"] + qualified["SB"] - 2 * qualified["CS"]
) / qualified["PA"]
```

```python
leaderboard = qualified.sort_values("SIMPSON", ascending=False)
columns = [column for column in ["Name", "Team", "1B", "SB", "CS", "PA", "SIMPSON"]
       if column in leaderboard.columns]
leaderboard[columns].head(10).round({"SIMPSON": 3})
```

```python
leaderboard[columns].to_csv("../leaderboards/simpsonv1_leaderboard.csv", index=False)
```
